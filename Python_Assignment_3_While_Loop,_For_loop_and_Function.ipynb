###2. Prompt the User:

  3, which represents the maximum number of

```python
attempts = 3
guess = int(input("Guess the secret number (1 to 10): "))
print("Attempts left:", attempts)
```

**Here the code ask the user to guess a number and set the maximum number of

**3.** **Implement the Guessing Logic:**

★ Use a while loop to allow the user to keep guessing until they get the

   correct number or run out of attempts.

★ Provide feedback to the user for each guess:

➢ If the guess is out of the valid range (1 to 10), inform the user.

➢ If the guess is greater than the secret number.

➢ If the guess is lower than the secret number.

➢ If the guess is correct, congratulate the user and end the game.


```python
import random
secret_number = random.randint(1, 10)
attempts = 3
while attempts > 0:
    guess = int(input("Guess the secret number (1 to 10): "))
    if guess == secret_number:
        print("Congratulations! You guessed the correct number.")
        break
    elif guess < secret_number:
        print("Try a higher number.")
    else:
          print("Try a lower number.")
    attempts -=1
    if attempts > 0:
        print("Attempts left:", attempts)
    else:
      print("Sorry, you have run out of attempts. the correct number was",
                  secret_number)
```

**This is a simple number guessing game. First, a random number
between 1 and 10 is generated and the user gets 3 attempts to guess it.
After each guess, the code check whether the number is correct, higher,
or lower than the secret_number.If the answer is wrong, one attempt is
reduced and the remaining attempts are shown.If the correct number is
guessed,break stops the loop**

4. **Control Statements:**

★ Use continue to skip the rest of the loop after informing the user if the
guess is out of range.

★ Use break to exit the loop when the guess is correct.

★ Use else with the while loop to provide a message like "Better luck next
time!" if the user runs out of attempts without guessing the correct number.


```python
import random
secret_number = random.randint(1, 10)
attempts = 3
while attempts > 0:
  guess=int(input("guess the number between 0 to 10:"))
  if guess < 1 or guess > 10:
    print("The guess is out of valid range,")
    continue
  if guess == secret_number:
    print("Congratulations! You guessed the correct number.")
    break
  elif guess < secret_number:
    print("try a higher number")
  else:
    print("try a lower number")
  attempts -= 1
else:
    print("Sorry, you have run out of attempts. Better luck next time!")
    print("The correct number was:", secret_number)
```

**The program gives the user 3 chances to guess the number. I used "Continue" when the entered number is outside 1 to 10, so the user can try again. I used "Break" to stop the game when the correct number is guessed. If the guess is wrong, the code tells whether to try a higher or lower number.**

# For Loop:

Multiplication Table Generator

Problem Statement:
 Create a Python program that generates and prints a
multiplication table (from 1 to 10) for a given number using a for loop and the range
function.

**Step wise Instructions**:
1. Prompt user for Input. Ask the user to enter a number for which they want to
generate a multiplication table.
2. Generate the Multiplication Table: Use a for loop to iterate through the numbers
1 to 10. In each iteration, calculate the product of the user's number and the
current number from the loop.
 3. Display the Multiplication Table:
 Print each line of the multiplication table in the format: "number x i = result".

```python
number = int(input("enter the number:"))
for i in range(1,11):
  result=number*i
  print(number, "x" ,i ,"=" ,result)
```

**I used a for loop to repeat the calculation 10 times. range(1, 11) gives the numbers from 1 to 10, and each number is multiplied by the number entered by the user.**
